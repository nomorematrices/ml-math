# Bài 19 — Expectation, Variance và Covariance

## Mục tiêu
Tính moment chính xác và ước lượng covariance từ samples.

```python
import numpy as np
values = np.array([0., 1., 2., 3.])
probabilities = np.array([.1, .2, .4, .3])
mean = values @ probabilities
variance = (values**2) @ probabilities - mean**2
print('mean/variance:', mean, variance)
```

## Bài tập
Sinh Gaussian 2D tương quan, tính covariance/correlation bằng công thức và NumPy. Kiểm tra symmetry, PSD và tác động của đổi đơn vị một feature.

```python
rng = np.random.default_rng(42)
target_cov = np.array([[1., .8], [.8, 2.]])
X = rng.multivariate_normal([0., 0.], target_cov, size=20_000)
centered = X - X.mean(axis=0)
cov = centered.T @ centered / (len(X)-1)
corr = cov / np.sqrt(np.outer(np.diag(cov), np.diag(cov)))
assert np.allclose(cov, cov.T)
assert np.all(np.linalg.eigvalsh(cov) >= -1e-12)
print(cov, corr, sep='\n')
# TODO: nhân feature 0 với 100 và so covariance/correlation
```
