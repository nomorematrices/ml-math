# Bài 17 — Random Variables và Distributions

## Mục tiêu
Tính joint, marginal và conditional distributions từ bảng xác suất.

```python
import numpy as np
joint = np.array([[0.10, 0.20, 0.10], [0.15, 0.25, 0.20]])
assert np.isclose(joint.sum(), 1)
p_x = joint.sum(axis=1)
p_y = joint.sum(axis=0)
p_x_given_y = joint / p_y[None, :]
print('p(x):', p_x, 'p(y):', p_y, 'p(x|y):', p_x_given_y, sep='\n')
np.testing.assert_allclose(p_x_given_y.sum(axis=0), 1)
```

## Bài tập
Mô phỏng 100.000 mẫu từ joint table, ước lượng joint/marginal và so với giá trị thật. Giải thích sai lệch lấy mẫu.

```python
rng = np.random.default_rng(42)
flat_samples = rng.choice(joint.size, size=100_000, p=joint.ravel())
empirical_joint = np.bincount(flat_samples, minlength=joint.size).reshape(joint.shape) / flat_samples.size
print(empirical_joint)
np.testing.assert_allclose(empirical_joint, joint, atol=0.01)
# TODO: suy ra empirical conditional distribution
```
