# Bài 20 — Các phân phối thường gặp

## Mục tiêu
Sinh mẫu và đối chiếu moments của năm distribution quan trọng.

```python
import numpy as np
rng = np.random.default_rng(42)
n = 100_000
bernoulli = rng.binomial(1, .3, n)
categorical = rng.choice(3, n, p=[.2, .5, .3])
normal = rng.normal(2., 1.5, n)
beta = rng.beta(2., 5., n)
dirichlet = rng.dirichlet([1., 2., 3.], n)
print(bernoulli.mean(), normal.mean(), beta.mean(), dirichlet.mean(axis=0))
```

## Bài tập
So sample mean/variance với lý thuyết. Vẽ histogram cho Normal/Beta và simplex scatter cho Dirichlet 3 chiều sau khi chiếu 2D.

```python
np.testing.assert_allclose(bernoulli.mean(), .3, atol=.01)
np.testing.assert_allclose(normal.var(), 1.5**2, atol=.05)
np.testing.assert_allclose(beta.mean(), 2/(2+5), atol=.01)
np.testing.assert_allclose(dirichlet.sum(axis=1), 1.)
counts = np.bincount(categorical, minlength=3)/n
np.testing.assert_allclose(counts, [.2, .5, .3], atol=.01)
# TODO: thay đổi alpha/beta và mô tả hình dạng Beta
```
