# Bài 18 — Bayes và Independence

## Mục tiêu
Cập nhật posterior và quan sát ảnh hưởng của base rate.

```python
import numpy as np
prevalence = 0.01
sensitivity = 0.95
specificity = 0.90
p_positive = sensitivity*prevalence + (1-specificity)*(1-prevalence)
posterior = sensitivity*prevalence/p_positive
print(f'P(disease | positive) = {posterior:.3%}')
```

## Bài tập
Quét prevalence từ 0.1% đến 50% và xem posterior thay đổi. Tạo joint table độc lập rồi phá vỡ một ô và kiểm tra `joint == outer(marginals)`.

```python
rates = np.array([.001, .01, .05, .1, .5])
posteriors = sensitivity*rates / (sensitivity*rates + (1-specificity)*(1-rates))
print(np.column_stack([rates, posteriors]))
p_x, p_y = np.array([.4, .6]), np.array([.2, .3, .5])
independent_joint = np.outer(p_x, p_y)
assert np.allclose(independent_joint, np.outer(independent_joint.sum(1), independent_joint.sum(0)))
# TODO: mô phỏng test y tế và so tần suất với Bayes
```
