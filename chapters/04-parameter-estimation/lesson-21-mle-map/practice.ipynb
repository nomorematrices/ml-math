# Bài 21 — MLE và MAP

## Mục tiêu
Ước lượng Bernoulli probability bằng MLE/MAP và phân tích ảnh hưởng của prior.

```python
import numpy as np
rng = np.random.default_rng(42)
data = rng.binomial(1, .7, size=20)
k, n = data.sum(), data.size
mle = k/n
alpha, beta = 2., 2.
map_estimate = (k+alpha-1)/(n+alpha+beta-2)
posterior_mean = (k+alpha)/(n+alpha+beta)
print({'successes': k, 'MLE': mle, 'MAP': map_estimate, 'posterior_mean': posterior_mean})
```

## Bài tập
Grid-search log-likelihood và log-posterior. So MLE/MAP khi n = 5, 20, 1000 và với prior yếu/mạnh. Tránh `log(0)`.

```python
grid = np.linspace(1e-4, 1-1e-4, 5000)
log_likelihood = k*np.log(grid) + (n-k)*np.log1p(-grid)
log_prior = (alpha-1)*np.log(grid) + (beta-1)*np.log1p(-grid)
grid_mle = grid[np.argmax(log_likelihood)]
grid_map = grid[np.argmax(log_likelihood + log_prior)]
np.testing.assert_allclose(grid_mle, mle, atol=5e-4)
np.testing.assert_allclose(grid_map, map_estimate, atol=5e-4)
print(grid_mle, grid_map)
# TODO: viết thí nghiệm theo cỡ mẫu và strength của prior
```
