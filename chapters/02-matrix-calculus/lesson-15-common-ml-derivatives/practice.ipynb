# Bài 15 — Đạo hàm thường gặp trong ML

## Mục tiêu
Suy ra least-squares gradient và dùng nó trong gradient descent.

```python
import numpy as np
rng = np.random.default_rng(42)
A = rng.normal(size=(30, 3))
true_x = np.array([2., -1., 0.5])
b = A @ true_x + 0.05*rng.normal(size=30)
def loss(x): return np.sum((A @ x - b)**2)
def gradient(x): return 2 * A.T @ (A @ x - b)
```

## Bài tập
Chạy gradient descent, lưu loss theo vòng lặp và so nghiệm với `np.linalg.lstsq`. Sau đó kiểm tra đạo hàm trace.

```python
x = np.zeros(3)
learning_rate = 0.01
history = []
for _ in range(500):
    history.append(loss(x))
    x -= learning_rate * gradient(x)
solution, *_ = np.linalg.lstsq(A, b, rcond=None)
np.testing.assert_allclose(x, solution, atol=1e-3)
assert history[-1] < history[0]
print(x, solution, history[-1], sep='\n')
# TODO: chứng minh grad_X trace(C @ X) = C.T bằng finite difference
```
