# Bài 14 — Product Rule và Chain Rule

## Mục tiêu
Thực hành chain rule như một computational graph nhỏ.

```python
import numpy as np
x = np.array([1., -2., 0.5])
a = np.array([2., 1., -1.])
u = a @ x
loss = u**2
d_loss_du = 2*u
d_u_dx = a
gradient = d_loss_du * d_u_dx
print('loss/gradient:', loss, gradient)
```

## Bài tập
Cho `loss = sum((sigmoid(A@x)-y)^2)`. Vẽ graph, tính từng local derivative và cài gradient theo x.

```python
A = np.array([[1., 2.], [-1., 0.5]])
x = np.array([0.2, -0.4])
y = np.array([1., 0.])
z = A @ x
s = 1 / (1 + np.exp(-z))
loss = np.sum((s-y)**2)
gradient = A.T @ (2*(s-y) * s*(1-s))
assert gradient.shape == x.shape
print(loss, gradient)
# TODO: kiểm tra gradient bằng central difference
```
