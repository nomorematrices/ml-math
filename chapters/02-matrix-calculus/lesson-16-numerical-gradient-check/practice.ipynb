# Bài 16 — Numerical Gradient Check

## Mục tiêu
Cài central difference, relative error và khảo sát step size.

```python
import numpy as np
def numerical_gradient(f, x, h=1e-5):
    result = np.zeros_like(x, dtype=float)
    for i in range(x.size):
        direction = np.zeros_like(x); direction[i] = h
        result[i] = (f(x+direction)-f(x-direction))/(2*h)
    return result
def relative_error(a, b):
    return np.linalg.norm(a-b)/max(np.linalg.norm(a)+np.linalg.norm(b), 1e-12)
```

## Bài tập
Kiểm tra gradient của `sum(exp(x)) + x.T@A@x`; quét `h=10^-1...10^-12`, tìm vùng sai số nhỏ nhất và giải thích đường cong chữ U.

```python
A = np.array([[2., 1.], [1., 3.]])
x = np.array([0.3, -0.7])
f = lambda z: np.sum(np.exp(z)) + z @ A @ z
analytic = np.exp(x) + (A + A.T) @ x
for h in 10.**(-np.arange(1, 13)):
    numerical = numerical_gradient(f, x, h)
    print(f'h={h:.0e}, error={relative_error(analytic, numerical):.3e}')
# TODO: thay analytic bằng công thức sai và quan sát error
```
