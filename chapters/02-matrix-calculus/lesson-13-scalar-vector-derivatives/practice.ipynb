# Bài 13 — Gradient và Jacobian

## Mục tiêu
Tính gradient/Jacobian và kiểm tra shape của đạo hàm.

```python
import numpy as np
x = np.array([1., -2., 3.])
gradient = 2 * x  # f(x) = x.T @ x
A = np.array([[1., 2., 0.], [0., -1., 3.]])
jacobian = A      # g(x) = A @ x + b
assert gradient.shape == x.shape
assert jacobian.shape == (2, 3)
print(gradient, jacobian, sep='\n')
```

## Bài tập
Với `f(x1,x2)=x1^2+3*x1*x2+sin(x2)`, tự suy ra gradient rồi so với finite difference.

```python
def f(z):
    return z[0]**2 + 3*z[0]*z[1] + np.sin(z[1])

point = np.array([0.5, -1.0])
analytic = np.array([2*point[0] + 3*point[1], 3*point[0] + np.cos(point[1])])
h = 1e-5
numerical = np.array([(f(point + h*np.eye(2)[i]) - f(point - h*np.eye(2)[i]))/(2*h) for i in range(2)])
np.testing.assert_allclose(analytic, numerical, rtol=1e-5)
# TODO: xây Jacobian cho [x1*x2, x1+x2^2]
```
