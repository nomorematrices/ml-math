# Bài 12 — Norm và Trace

## Mục tiêu
So sánh vector norms, Frobenius norm và các identity của trace.

```python
import numpy as np
x = np.array([1., -2., 3.])
for order in [1, 2, 3, np.inf]:
    print(order, np.linalg.norm(x, ord=order))
A = np.array([[1., 2.], [3., 4.]])
frobenius = np.linalg.norm(A, ord='fro')
np.testing.assert_allclose(frobenius**2, np.trace(A.T @ A))
```

## Bài tập
Kiểm tra triangle inequality trên 1000 cặp vector ngẫu nhiên. Kiểm chứng `trace(ABC)=trace(BCA)` với shape tương thích.

```python
rng = np.random.default_rng(42)
u, v = rng.normal(size=(2, 1000, 5))
lhs = np.linalg.norm(u + v, axis=1)
rhs = np.linalg.norm(u, axis=1) + np.linalg.norm(v, axis=1)
assert np.all(lhs <= rhs + 1e-12)
B, C = rng.normal(size=(2, 2, 2))
np.testing.assert_allclose(np.trace(A @ B @ C), np.trace(B @ C @ A))
# TODO: so sánh hình học L1/L2/L-infinity
```
