# Bài 11 — Positive Definite và Cholesky

## Mục tiêu
Kiểm tra PD/PSD và dùng Cholesky trong tính toán.

```python
import numpy as np
A = np.array([[4., 2.], [2., 3.]])
eigenvalues = np.linalg.eigvalsh(A)
L = np.linalg.cholesky(A)
print('eigenvalues:', eigenvalues)
np.testing.assert_allclose(L @ L.T, A)
assert np.all(eigenvalues > 0)
```

## Bài tập
Tạo `X.T @ X`, kiểm tra PSD; thêm `lambda * I`, kiểm tra PD; dùng Cholesky sinh Gaussian có covariance chỉ định.

```python
rng = np.random.default_rng(42)
X = rng.normal(size=(3, 5))
gram = X.T @ X
regularized = gram + 0.1 * np.eye(gram.shape[0])
assert np.all(np.linalg.eigvalsh(gram) > -1e-12)
assert np.all(np.linalg.eigvalsh(regularized) > 0)
samples = rng.normal(size=(20000, 2)) @ L.T
np.testing.assert_allclose(np.cov(samples, rowvar=False), A, atol=0.12)
# TODO: giải thích vai trò của regularization
```
