# Bài 10 — Eigendecomposition

## Mục tiêu
Tái tạo ma trận từ eigenpairs và dùng phân rã để tính lũy thừa.

```python
import numpy as np
A = np.array([[4., 1.], [1., 2.]])
values, Q = np.linalg.eigh(A)
Lambda = np.diag(values)
reconstructed = Q @ Lambda @ Q.T
np.testing.assert_allclose(reconstructed, A)
print('reconstruction error:', np.linalg.norm(A-reconstructed))
```

## Bài tập
Tính `A^20` trực tiếp và qua `Q @ Lambda^20 @ Q.T`. Sau đó thử ma trận defective `[[1,1],[0,1]]` và kiểm tra rank eigenvectors.

```python
power_direct = np.linalg.matrix_power(A, 20)
power_eigen = Q @ np.diag(values**20) @ Q.T
np.testing.assert_allclose(power_direct, power_eigen, rtol=1e-10)
defective = np.array([[1., 1.], [0., 1.]])
_, V = np.linalg.eig(defective)
print('eigenvector rank:', np.linalg.matrix_rank(V))
# TODO: giải thích vì sao V không khả nghịch về mặt lý thuyết
```
