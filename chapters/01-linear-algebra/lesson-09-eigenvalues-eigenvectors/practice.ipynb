# Bài 09 — Eigenvalue và Eigenvector

## Mục tiêu
Tính, sắp xếp và kiểm tra eigenpairs của ma trận đối xứng.

```python
import numpy as np
A = np.array([[3., 1.], [1., 3.]])
values, vectors = np.linalg.eigh(A)
order = np.argsort(values)[::-1]
values, vectors = values[order], vectors[:, order]
for value, vector in zip(values, vectors.T):
    np.testing.assert_allclose(A @ vector, value * vector)
print(values, vectors, sep='\n')
```

## Bài tập
Sinh dữ liệu 2D tương quan, tính covariance matrix và tìm hướng eigenvector chính. So sánh variance sau khi chiếu lên hai eigenvector.

```python
rng = np.random.default_rng(42)
z = rng.normal(size=500)
X = np.column_stack([z + 0.1*rng.normal(size=500), 2*z + 0.2*rng.normal(size=500)])
cov = np.cov(X, rowvar=False)
eigenvalues, eigenvectors = np.linalg.eigh(cov)
principal = eigenvectors[:, -1]
print('principal direction:', principal)
# TODO: tính variance của X @ eigenvectors và so với eigenvalues
```
